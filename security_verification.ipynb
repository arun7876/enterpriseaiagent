```python
import json
import shutil
import subprocess
from pathlib import Path

project = Path(r'C:\Users\ARUNACHALAM\OneDrive\Desktop\Enterprise ai-agent\enterprise-ai-agent')
python = str(project / '.venv' / 'Scripts' / 'python.exe')

checks = [
    ('backend_full', [python, '-m', 'pytest', '-q'], project),
    ('config_validation', [python, '-c', 'from backend.app.config import settings; settings.validate(); print("configuration valid")'], project),
]
npm = shutil.which('npm.cmd') or shutil.which('npm')
if npm:
    npm_prefix = ['cmd.exe', '/c', npm] if npm.lower().endswith('.cmd') else [npm]
    checks.extend([
        ('frontend_ci_install', npm_prefix + ['ci'], project / 'frontend'),
        ('frontend_tests', npm_prefix + ['test'], project / 'frontend'),
        ('frontend_lint', npm_prefix + ['run', 'lint'], project / 'frontend'),
        ('frontend_build', npm_prefix + ['run', 'build'], project / 'frontend'),
    ])
else:
    print('NPM_NOT_FOUND')

for name, command, cwd in checks:
    outcome = subprocess.run(command, cwd=cwd, capture_output=True, text=True)
    output = (outcome.stdout + '\n' + outcome.stderr).strip()
    print(f'--- {name}: exit {outcome.returncode} ---')
    print('\n'.join(output.splitlines()[-8:]))

print('--- availability ---')
print('docker', shutil.which('docker'))
print('git', shutil.which('git'))
if shutil.which('git'):
    inside = subprocess.run(['git', 'rev-parse', '--is-inside-work-tree'], cwd=project, capture_output=True, text=True)
    print('git_repo', inside.stdout.strip() if inside.returncode == 0 else 'not initialized')
    history = subprocess.run(['git', 'log', '-1', '--oneline'], cwd=project, capture_output=True, text=True)
    print('git_history', history.stdout.strip() if history.returncode == 0 else 'no commits')
try:
    import yaml
    for name in ['docker-compose.yml', '.github/workflows/ci.yml']:
        yaml.safe_load((project / name).read_text(encoding='utf-8'))
        print('yaml_valid', name)
except Exception as exc:
    print('yaml_validation_error', type(exc).__name__, str(exc))
```

```python
import json
import shutil
import subprocess
from pathlib import Path

project = Path(r'C:\Users\ARUNACHALAM\OneDrive\Desktop\Enterprise ai-agent\enterprise-ai-agent')
python = str(project / '.venv' / 'Scripts' / 'python.exe')
def run(name, command, cwd):
    result = subprocess.run(command, cwd=cwd, capture_output=True, text=True, encoding='utf-8', errors='replace')
    output = ((result.stdout or '') + '\n' + (result.stderr or '')).strip()
    print(f'--- {name}: exit {result.returncode} ---')
    print('\n'.join(output.splitlines()[-10:]))
    return result

run('backend_full_suite', [python, '-m', 'pytest', '-q'], project)
run('security_suite', [python, '-m', 'pytest', '-q', 'tests/test_auth_security.py'], project)
run('configuration_validation', [python, '-c', 'from backend.app.config import settings; settings.validate(); print("configuration valid")'], project)
npm = shutil.which('npm.cmd') or shutil.which('npm')
prefix = ['cmd.exe', '/c', npm] if npm and npm.lower().endswith('.cmd') else [npm]
for name, args in [('frontend_tests', ['test']), ('frontend_lint', ['run', 'lint']), ('frontend_build', ['run', 'build'])]:
    run(name, prefix + args, project / 'frontend')
print('--- runtime versions ---')
print('python', subprocess.run([python, '--version'], capture_output=True, text=True).stdout.strip())
print('node', subprocess.run(['node', '--version'], capture_output=True, text=True).stdout.strip())
print('npm', subprocess.run(prefix + ['--version'], capture_output=True, text=True, encoding='utf-8', errors='replace').stdout.strip())
print('docker', shutil.which('docker'))
try:
    import yaml
    for name in ['docker-compose.yml', '.github/workflows/ci.yml']:
        yaml.safe_load((project / name).read_text(encoding='utf-8'))
        print('yaml_valid', name)
except Exception as exc:
    print('yaml_validation_error', type(exc).__name__, str(exc))
print('--- git ---')
git = shutil.which('git')
if git:
    inside = subprocess.run([git, 'rev-parse', '--is-inside-work-tree'], cwd=project, capture_output=True, text=True)
    print('repo', inside.stdout.strip() if inside.returncode == 0 else 'not initialized')
    history = subprocess.run([git, 'log', '-1', '--oneline'], cwd=project, capture_output=True, text=True)
    print('history', history.stdout.strip() if history.returncode == 0 else 'no commits')
    remotes = subprocess.run([git, 'remote'], cwd=project, capture_output=True, text=True)
    print('remote_names', remotes.stdout.strip() or 'none')
    status = subprocess.run([git, 'status', '--short', '--untracked-files=all'], cwd=project, capture_output=True, text=True)
    lines = status.stdout.strip().splitlines()
    print('status_entries', len(lines))
    print('\n'.join(lines[:45]))
    ignored = subprocess.run([git, 'check-ignore', '-v', '.env', 'data/auth.sqlite3', 'data/documents.sqlite3', 'data/uploads/private.csv', 'data/index/vector_index.sqlite3', 'data/models/huggingface/model.safetensors'], cwd=project, capture_output=True, text=True)
    print('ignored_private_paths', ignored.stdout.strip())
```
